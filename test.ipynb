```python
fig, ax = plt.subplots()
t = np.linspace(0, 3, 40)
t0 = 2
g = -9.81
v0 = 12
z = g * t**2 / 2 + v0 * t

v02 = 5
z2 = g * t**2 / 2 + v02 * t

scat = ax.scatter(t[0], z[0], c="b", s=5, label=f'v0 = {v0} m/s')
line2 = ax.plot(t[0], z2[0], label=f'v0 = {v02} m/s')[0]
ax.set(xlim=[0, 3], ylim=[-4, 10], xlabel='Time [s]', ylabel='Z [m]')
ax.legend()


def update(frame):
    # for each frame, update the data stored on each artist.
    x = np.linspace(t0-.5, t0+.5, 10)
    y = z[frame] * np.ones_like(x)

    # update the scatter plot:
    # data = np.stack([x, y]).T
    # scat.set_offsets(data)
    # print(data)
    # update the line plot:
    line2.set_xdata(x)
    line2.set_ydata(y)
    return (scat, line2)


ani = animation.FuncAnimation(fig=fig, func=update, frames=40, interval=30)
ani.save(fr'C:\\Users\\tommy\\OneDrive - Durham University\\level 4 Project\\anim.gif')

plt.show()
```

```python
import pairinteraction as pi

# 1. Define the atoms
atom = pi.Atom("Rb")

# 2. Define two specific pair-state "tuples"
# State 1: (60S, 60S)
s1 = pi.State(atom, 60, 0, 0.5, 0.5)
pair_tuple_1 = pi.StatePair(s1, s1)

# State 2: (60P, 60P)
s2 = pi.State(atom, 60, 1, 1.5, 1.5)
pair_tuple_2 = pi.StatePair(s2, s2)

# 3. Build a BasisPair from ONLY these two tuples
# This locks the Hilbert space to size 2
custom_basis = pi.BasisPair([pair_tuple_1, pair_tuple_2])

# 4. Initialize EffectiveSystemPair with this custom basis
# Note: You still pass the 'target' tuple as the first argument
system = pi.EffectiveSystemPair(pair_tuple_1, pair_tuple_1, custom_basis)

# 5. Extract the 2x2 Hamiltonian
system.set_distance(10.0) # distance in micrometers
hamiltonian = system.get_hamiltonian()

print(f"Hamiltonian shape: {hamiltonian.shape}") # Should be (2, 2)
print(hamiltonian)
```
